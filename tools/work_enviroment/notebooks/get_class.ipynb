```python
import numpy as np
import torch
import matplotlib.pyplot as plt
import cv2
import os
import sys
from segment_anything import sam_model_registry, SamAutomaticMaskGenerator, SamPredictor
from PIL import Image
sys.path.append("..")
```

```python
sam_checkpoint = "sam_vit_h_4b8939.pth"
model_type = "vit_h"
device = "cuda"
```

```python
sam = sam_model_registry[model_type](checkpoint=sam_checkpoint)
sam.to(device=device)
mask_generator = SamAutomaticMaskGenerator(sam)
```

```python
data_image_folder = '/torch/data/'
data_image_names = [os.path.join(data_image_folder, i) for i in os.listdir(data_image_folder)]
```

```python
for num, image_name in enumerate(data_image_names[499:]):
    torch.cuda.empty_cache()
    image = cv2.imread(image_name)
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    masks = mask_generator.generate(image)
    for n, m in enumerate(masks):
        m_arr = m['segmentation'].astype('uint8')
        m_img = cv2.merge([m_arr, m_arr, m_arr])
        res_mask = cv2.multiply(image, m_img)
        fn_mask = os.path.join(os.path.join(data_image_folder, 'masks'), os.path.split(image_name)[-1].replace('.png', f'_{n}.png'))
        im_mask = Image.fromarray(np.array(res_mask))
        im_mask.save(fn_mask)
    print(num, image_name)
```
